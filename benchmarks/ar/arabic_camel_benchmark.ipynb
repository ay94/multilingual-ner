```python
from multilingual_ner.evaluation import ReadNERData, ModelEvaluation, align_dataset, check_labels
```

## Load dataset

```python
label_map = {'O': 0, 'B-PER': 1, 'I-PER': 2, 'B-ORG': 3, 'I-ORG': 4, 'B-LOC': 5, 'I-LOC': 6}

reader = ReadNERData()
wikiann_words, wikiann_labels = reader.read_dataset('wikiann', label_map, lang='ar')
print(f'Loaded {len(wikiann_words)} sentences')
print('Labels:', check_labels(wikiann_labels))
```

## Load model & define alignment

CAMeL uses `person / location / organization / other` labels — align to standard BIO scheme.

```python
model_alignment = {
    'B-person':       'B-PER', 'I-person':       'I-PER',
    'B-location':     'B-LOC', 'I-location':     'I-LOC',
    'B-organization': 'B-ORG', 'I-organization': 'I-ORG',
    'B-other':        'O',     'I-other':        'O',
    'O':              'O',
}

model = ModelEvaluation('CAMeL-Lab/bert-base-arabic-camelbert-mix-ner', model_alignment)
print(model.model.config.id2label)
```

## Evaluate — WikiANN Arabic

```python
results_model_wikiann = model.evaluate_model(wikiann_words, wikiann_labels)

print('--- Seqeval (entity-level) ---')
display(results_model_wikiann.get_classification('Seqeval'))

print('--- Sklearn (token-level) ---')
display(results_model_wikiann.get_classification('Sklearn'))
```
