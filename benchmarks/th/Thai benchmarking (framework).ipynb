```python
from ner import list_models, Language, License
from ner.languages.th.datasets.ThaiNER import ThaiNER
from ner.languages.th.datasets.wikiann import Wikiann
```

```python
datasets = [ThaiNER(), Wikiann()]
for model_class in list_models(Language.THAI, License):
    model = model_class()
    for dataset in datasets:
        print(f"Model: {model.model_name}, Dataset: {dataset.dataset_name}")
        print(model.seqeval(dataset, test=True))  # REMOVE TEST FOR FINAL EVAL
```
